# Recording Audio Data

```python
import matplotlib.pyplot as plt
import numpy as np
import sfs
import soundfile as sf
```

```python
f = 300  # TODO: where are artifacts audible?
amp = 0.3
input_duration = 2  # seconds
fs = 48000
```

```python
input_t = np.arange(int(input_duration * fs)) / fs
```

Here we are creating a (mono) sine signal,
but we could just as well generate any other signal
or load one from an audio file.

```python
input_signal = amp * np.sin(2 * np.pi * f * input_t)
```

```python
# TODO: fade in/out
```

```python
# TODO: plot sound field (zoomed in to see waves)
```

```python
xs = 0, 0, 0  # sound source position / m
pos1 = 0, 1, 0
pos2 = 1, 1, 0

positions = [
    (0, 1, 0),
    (1, 1, 0),
]
```

```python
output_duration = 3  # TODO: tighter fit
```

```python
output_t = np.arange(int(output_duration * fs)) / fs
```

TODO: we normalize to +-1 Pascal at reference position

```python
normalizing_factor = 4 * np.pi * 1  # TODO: use min(x1, x2) to get +-1 amplitude?
```

```python
p = normalizing_factor * sfs.td.source.point(
    xs, (input_signal, fs), observation_time=output_t, grid=zip(*positions))
```

```python
sf.write('output01.wav', p, fs)  # TODO: less ambiguous file name
```
